```python
curr_index = 290
WT_seq = "G"
curr_aa = "G"
for i in single_mutants.index:
    if curr_index == int(single_mutants.loc[i, "Pos_abs"]):
        if curr_aa == single_mutants.loc[i, "WT_AA"]:
            continue
        else:

            print("ERROR 1")
    elif curr_index + 1 == int(single_mutants.loc[i, "Pos_abs"]):
        WT_seq += single_mutants.loc[i, "WT_AA"]
        curr_index += 1
        curr_aa = single_mutants.loc[i, "WT_AA"]
        # print(curr_index, WT_seq)
    else:
        print(curr_index, int(single_mutants.loc[i, "Pos_abs"]))
        print("ERROR 2")

```

```python
single_mutants['aa_seq'] = [WT_seq[:int(single_mutants.loc[i, 'Pos']) - 1] + single_mutants.loc[i, 'Mut'] + WT_seq[int(single_mutants.loc[i, 'Pos']):] if not np.isnan(single_mutants.loc[i, 'Pos']) else WT_seq for i in single_mutants.index]
double_mutants['aa_seq'] = [WT_seq[:int(double_mutants.loc[i, 'Pos1']) - 1] + double_mutants.loc[i, 'Mut1'] + WT_seq[int(double_mutants.loc[i, 'Pos1']):int(double_mutants.loc[i, 'Pos2'])] + double_mutants.loc[i, 'Mut2'] + WT_seq[int(double_mutants.loc[i, 'Pos2']):] for i in double_mutants.index]
```

```python
double_mutants['toxicity'] = double_mutants['toxicity_cond']
```

```python
WT_avg_df = pd.DataFrame(WT_seqs.mean()).T
WT_avg_df['aa_seq'] = WT_seq
```

```python
all_mutants = pd.concat([single_mutants, double_mutants, WT_avg_df])
```
